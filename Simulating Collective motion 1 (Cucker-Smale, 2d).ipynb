```python
import pygame
import random
import math

# Initialize Pygame
pygame.init()

# Parameters i will adjust
rad = 30
num = 300
eta = 0.01*math.pi
alpha = 0.08
# Screen settings
WIDTH = 1200
HEIGHT = 800
FPS = 30
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Particle Simulation with Wrapping")
clock = pygame.time.Clock()

# Color constants
BLACK = (0, 0, 0)
WHITE = (240, 240, 240)
RED = (255, 0, 0)
BLUE = (150,150,240)
YELLOW = (240,240,180)
# Particle class
class Particle:
    def __init__(self, x, y, angle):
        self.x = x
        self.y = y
        # Random velocity direction (constant speed)
        self.angle = angle
        self.v = 4 # Constant speed
        self.radius = 3
    
    def update(self, all_particles, rad):
        """Update particle position and wrap around edges"""
        # Thi is for clipping
        if self.x < 0:
            self.x += WIDTH
        elif self.x > WIDTH:
            self.x -= WIDTH

        if self.y < 0:
            self.y += HEIGHT
        elif self.y > HEIGHT:
            self.y -= HEIGHT

        #This is for updating movement
        self.x += self.v * math.cos(self.angle)
        self.y += self.v * math.sin(self.angle)

        #This take average velocit of all particles
        neighbors = set()
        for other in all_particles:
            if other == self:
                continue
            distance = math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)
            if distance < rad:
                neighbors.add(other)
        # UPdates particles with the verage
        if len(neighbors) == 0:
            return
        sin_sum = sum(math.sin(n.angle) for n in neighbors) / len(neighbors)
        cos_sum = sum(math.cos(n.angle) for n in neighbors) / len(neighbors)
        avg_angle = math.atan2(sin_sum, cos_sum)
        noise = random.uniform(-eta, eta)
        self.angle = self.angle * (1-alpha) + avg_angle * alpha  + noise

        
        # This causes the edge to lead into eachother
    
    def draw(self, screen):
        # Create arrow points (triangle)
        # Point 1: tip of arrow
        tip_x = self.x + self.radius * 2 * math.cos(self.angle)
        tip_y = self.y + self.radius * 2 * math.sin(self.angle)
    
        # Point 2: left back corner
        left_x = self.x + self.radius * 1.5 * math.cos(self.angle + 2.5)
        left_y = self.y + self.radius * 1.5 * math.sin(self.angle + 2.5)
    
        # Point 3: right back corner
        right_x = self.x + self.radius * 1.5 * math.cos(self.angle - 2.5)
        right_y = self.y + self.radius * 1.5 * math.sin(self.angle - 2.5)
    
        # Draw triangle
        points = [(tip_x, tip_y), (left_x, left_y), (right_x, right_y)]
        pygame.draw.polygon(screen, YELLOW, points)


particles = [Particle(random.uniform(0, WIDTH), random.uniform(0, HEIGHT), random.uniform(-4*math.pi,4*math.pi)) for _ in range(num)]

# Main simulation loop
running = True
while running:
    clock.tick(FPS)
    
    # Event handling
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
    
    # Update all particles
    for particle in particles:
        particle.update(particles, rad)
    
    # Draw
    screen.fill(BLUE)  # Clear screen
    for particle in particles:
        particle.draw(screen)
    
    pygame.display.flip()  # Update display

pygame.quit()
```
